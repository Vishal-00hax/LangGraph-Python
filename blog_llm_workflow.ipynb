```python
import os
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()  # Load environment variables from .env file
```

```python
class BlogState(TypedDict):
    title: str
    outline: str
    content: str
```

```python
model = ChatOpenAI(
    model=os.getenv("GEMINI_AI_MODEL"),
    api_key=os.getenv("GEMINI_API_KEY"),
    base_url="https://generativelanguage.googleapis.com/v1beta/openai/",
    temperature=0
)
```

```python
def create_blog_outline(state: BlogState) -> BlogState:
    prompt = f"Create a detailed outline for a blog post titled '{state['title']}'."
    outline = model.invoke(prompt).content
    state["outline"] = outline
    return state
```

```python
def create_blog_content(state: BlogState) -> BlogState:
    prompt = f"Write a blog post based on the following outline:\n{state['outline']} \n\n Title: {state['title']}"
    content = model.invoke(prompt).content
    state["content"] = content
    return state
```

```python
graph = StateGraph(BlogState)
graph.add_node("create_outline", create_blog_outline)
graph.add_node("create_content", create_blog_content)
graph.add_edge(START, "create_outline")
graph.add_edge("create_outline", "create_content")
graph.add_edge("create_outline", END)

workflow = graph.compile()
```

```python
# Execute the workflow (graph)
inital_state = {"title": "AI in Education"}
output_state = workflow.invoke(inital_state)
print(output_state["content"])
```
